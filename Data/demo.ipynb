```python
import json
```

```python
def load_data(file_name):
    with open(file_name, "r") as f:
        py_obj = json.load(f)
        
    return py_obj
```

```python
data = load_data("store_data.json")
print(data)
```

```python
def clean_data(data):
    string_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()
    for user in data:
        # clean ratings 
        raw_ratings = str(user["rating"]).strip().lower()    # strip() cut the extra spaces
        if(raw_ratings in string_to_num):
            raw_ratings = string_to_num[raw_ratings]
        user["rating"] = raw_ratings
        # print(user["rating"])

        # handle missing value

        raw_age = user.get("age")    # using get function(instead of user["age"] to access the key bcz if key is not present then it doesn't give error(it return none)
        if(raw_age == None):
            user["age"] = None

        #  remove duplicates (deduplication)
        if(user["name"].strip() in unique_users):
            continue
        else:
            cleaned_data.append(user)
            unique_users.add(user["name"].strip())
    return cleaned_data
```

```python
data = clean_data(data)
print(data)
```

```python
def ratings_info(data):
    print(data)
    total_user = len(data)
    rating_sum = 0
    poor_rating = 0

    # Avg rating
    for user in data:
        rating_sum+=float(user["rating"])
    print(f"Avg rating = {rating_sum/total_user}")

    # poor rating percentage
    for user in data:
        if(float(user["rating"])<3):
            poor_rating+=1
    print(f"Poor ratings% = {(poor_rating/total_user)*100}")

    
```

```python
ratings_info(data)
```

```python
#  Build a recomendation feature
def get_recommendation(data):
    recommendation = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]
        if(float(user["rating"])>=4): 
            curr_recomm["brand"] = "Apple"
        else:
            
            curr_recomm["brand"] = "Samsung"

        recommendation.append(curr_recomm)

    return recommendation
            
```

```python
get_recommendation(data)
```
