```python
import numpy as np
import time
```

```python
# np arr are faster than py list so use use np arr
arr = np.array([1,2,5,4,6])
print(arr, type(arr), arr.dtype, arr.shape)

arr2 = np.array([1,2,3,4,"hii"])      #list can contain diff data types but arr can't so if we take a str in the li to convert to arr so arr converts all el to str
print(arr2, type(arr2), arr2.dtype, arr2.shape)
```

```python
# comparing execution performance
size = 1_000_000     #size of li and nparr
py_list = list(range(size))     # creating li of 1M size contains data from 1 to 1M
np_arr = np.array(py_list)      # creating nparr of 1M with same data of py list

start = time.time()     # time.time() gives curr time
sq_list = [x**2 for x in py_list]
end = time.time()
print(f"Execution time = {end-start}")

start = time.time()
sq_arr = np_arr**2      # Vectorisation - In list loop is used but in arr it is directly done by vectorized.it performs operation on all el simultaneously and don't travese each el. 
end = time.time()
print(f"Execution time = {end-start}")
```

```python
# memory comparison - list take more space. it is bcz li el are not stored in contigious memory so it has to kept extra meta data to keep this info.
import sys

print(f"size of py_list = {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f"size of np_arr = {np_arr.nbytes} bytes")
```
