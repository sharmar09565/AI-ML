```python
import numpy as np
```

```python
# mathematical functions

# aggregation func
arr = np.array([1,2,3,4,5])

print(np.sum(arr))
print(np.prod(arr))
print(np.max(arr))
print(np.min(arr))

print(np.argmax(arr))   #return idx of max val
print(np.argmin(arr))   #return idx of min val

print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))      # return standard deviation
print(np.var(arr))      # return variance(sq of std or sigma sq)
```

```python
# power func

print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr, 3))
```

```python
# log

print(np.log(arr))      # here base is e
print(np.log10(arr))    # here base is 10
print(np.log2(arr))     # here base is 2
```

```python
# rounding
print(np.round(3.14))
print(np.ceil(3.14))    # round up
print(np.floor(3.14))   # round down
print(np.trunc(3.14))

nums = np.array([1,2,3,1,5,6,3])

print(np.unique(nums))      # give unique val, and does not repeat the same el
print(np.sort(nums))

n = np.array([1,-2,3,-1,-5])
print(np.abs(n))    # converts all the arr el into +ve
```
