```python
import pandas as pd
```

```python
# Merging and Joining data

# inner join
# left join
# right join

df_customers = pd.DataFrame({
    'customer_id': [1,2,3,4],
    'name': ['Adam','Bob','Charlie','Dave']
})

df_orders = pd.DataFrame({
    'order_id': [101,102,103,104],
    'customer_id': [2,1,4,5],
    'amount': [250,120,300,100]
})

pd.merge(df_customers, df_orders, on='customer_id')  # inner join

pd.merge(df_customers, df_orders, on='customer_id', how='left')     # take all the value of left side and replace the right val with Nan if left side val is not present in right side

pd.merge(df_customers, df_orders, on='customer_id', how='right')

pd.merge(df_customers, df_orders, on='customer_id', how='outer')    # comes the all val of both left and right
```
