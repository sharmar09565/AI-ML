```python
import pandas as pd
```

```python
# concatenate - we can add data side by side(cols increase) or on the top of each other(rows increases)

df1 = pd.DataFrame({
    'id':[1,2,3],
    'name':['Ram','Shyam','Prem']
})

df2 = pd.DataFrame({
    'id':[4,5,6],
    'name':['Vinod','Ajay','Ganesh']
})

pd.concat([df1,df2])    # row
pd.concat([df1,df2], ignore_index=True)

pd.concat([df1,df2], axis=1)
```
