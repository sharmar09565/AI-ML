```python
import pandas as pd
```

```python
#  Transforming data (Feature Engineering) - means which feature we have to select or add

# (a) apply()       // when write a transform logic inside a func then we use apply func under which pass that func as para
# (b) map()         // change the existing feature
# (c) assign()      // create new policy
# (d) replace(old,new)

df = pd.read_csv('raw_data.csv')

df2 = df.copy()
df2
df2['tax'] = df2['income'].apply(lambda x: "20%" if x>=50_000 else '10%')

gender_map = {'Male':'M', 'Female':'F', 'Unknown':'U'}
df2['gender'] = df2['gender'].map(gender_map)

df2 = df2.assign(new_income = df2['income']*2)

df2['country'] = df2['country'].replace('USA','US')
df2
```

```python
# some more methods
df2.columns = ['ID','Name','Age','Country','Gender','Income','Tax','New_Income'] # change all the columns name

df2.rename(columns={'Income':'Salary'})
# df2.rename(index={1:'First'})
df2.sort_values('Income')
df2.sort_values('Income', ascending=False)

df2.sort_values(['Income','Age'])     # if income is same then it checks age

# reset
sorted_df = df2.sort_values('Income')

sorted_df.reset_index()     # brings new idx starts from 0 but old idx is preserved
sorted_df.reset_index(drop=True)    # old idx is removed now

# ranking
sorted_df['Ranking'] = sorted_df['Income'].rank()   # ties of or more val takes the average so rank may comes in float
sorted_df['Ranking'] = sorted_df['Income'].rank(method='dense')    # can also use min and max instead of dense 

# reorder
sorted_df[['ID','Name','Age','Country','Gender','Income','New_Income','Tax']]
```
