```python
import pandas as pd
```

```python
df = pd.read_csv('raw_data.csv')

# grouping and aggregation of data
# (a) groupby()     // on which do you want group and calculate(like mean,add,max,min) which col. We apply groupby and then aggregate  
# (b) agg()     // when we want to aggregate multiple things together (like sum and mean)

# groupby()
df.groupby('country')['income'].mean()      # country avg income
df.groupby('country')['income'].min()  
df.groupby('country')['income'].max()  
df.groupby('country')['income'].sum()    

df.groupby('gender')['income'].mean()        # gender avg income
df.groupby('gender')['income'].sum()


# agg() - we can also use aggregate()
df.groupby('country')['income'].agg(['mean','sum','min','max']) 

df.groupby('country')['income'].agg(avg_salary='mean',min_salary='min',max_salary='max') 

#  we can also work on multiple col together
df.groupby('country').agg({
    'income': 'max',
    'age': 'mean'
})

# can also rename the col

df.groupby('country').agg(
    max_salary = ('income', 'max'),
    avg_age = ('age', 'mean')
)
```
