```python
# vectorize operation
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([10,20,30,40,50])

print(s1+s2)

s1[0] = 100 
changed_s1 = s1.drop(0)     # we can't add and remove el from original series. We have take new var if we want to add or remove the data 

print(s1)
print(changed_s1)
```
